```python
import geoglows
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Define the site code
site_code = '11160500'

# Get the historical streamflow data
historical_data = geoglows.streamflow.historic_simulation(site_code)

# Convert the data to a pandas DataFrame
historical_df = pd.DataFrame(historical_data)

# Print the first few rows of the data
print(historical_df.head())

# Plot the historical streamflow data
plt.figure(figsize=(12, 6))
plt.plot(historical_df['datetime'], historical_df['streamflow'], label='Streamflow')
plt.xlabel('Date')
plt.ylabel('Streamflow (m3/s)')
plt.title(f'Streamflow for Site {site_code}')
plt.legend()
plt.grid()
plt.show()
```

```python
# Get the forecast streamflow data
forecast_data = geoglows.streamflow.forecast_stats(site_code)
forecast_df = pd.DataFrame(forecast_data)
print(forecast_df.head())

plt.figure(figsize=(12, 6))
plt.plot(forecast_df['datetime'], forecast_df['mean'], label='Mean Forecast')
plt.fill_between(forecast_df['datetime'], forecast_df['min'], forecast_df['max'], color='gray', alpha=0.2, label='Min-Max Range')
plt.xlabel('Date')
plt.ylabel('Streamflow (m3/s)')
plt.title(f'Forecast Streamflow for Site {site_code}')
plt.legend()
plt.grid()
plt.show()
```
