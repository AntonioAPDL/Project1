```r
library(Rcpp)
library(moments)

# Compile the C++ code
# sourceCpp("/home/jaguir26/projects/notebooks/gig_test.cpp")

# Set parameters for GIG distribution
p <- 0.5
a <- 1.0
b_vec <- c(0.5, 1.0, 2.0)  # Vector of b values
n_samples <- 100000

# Generate samples
samples <- sample_gig_devroye_vector(n_samples, p, a, b_vec)

# Calculate empirical moments for each column
for (t in 1:length(b_vec)) {
  cat("For b =", b_vec[t], ":\n")
  empirical_mean <- mean(samples[, t])
  empirical_variance <- var(samples[, t])
  empirical_skewness <- skewness(samples[, t])
  empirical_kurtosis <- kurtosis(samples[, t])
  
  omega <- sqrt(a * b_vec[t])
  Kp <- besselK(sqrt(a * b_vec[t]), p)
  Kp1 <- besselK(sqrt(a * b_vec[t]), p + 1)
  Kp2 <- besselK(sqrt(a * b_vec[t]), p + 2)
  
  theoretical_mean <- sqrt(b_vec[t]) * Kp1 / (sqrt(a) * Kp)
  theoretical_variance <- (b_vec[t] / a) * ((Kp2 / Kp) - (Kp1 / Kp)^2)
  theoretical_mode <- ((p - 1) + sqrt((p - 1)^2 + a * b_vec[t])) / a
  
  cat("  Empirical Mean: ", empirical_mean, "\n")
  cat("  Theoretical Mean: ", theoretical_mean, "\n")
  cat("  Empirical Variance: ", empirical_variance, "\n")
  cat("  Theoretical Variance: ", theoretical_variance, "\n")
  cat("  Empirical Skewness: ", empirical_skewness, "\n")
  cat("  Empirical Kurtosis: ", empirical_kurtosis, "\n")
  cat("  Theoretical Mode: ", theoretical_mode, "\n\n")
}
```
