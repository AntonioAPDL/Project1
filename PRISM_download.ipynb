```r
.libPaths(c("~/R/libs", .libPaths()))
print(.libPaths())
library(dplyr)
library(prism)
library(raster)
library(sp)

prism_set_dl_dir("/data/muscat_data/jaguir26/project1_ucsc_phd/prism_data")
# Download data (your download already completed, so this might skip again)
get_prism_dailys(type = "ppt", minDate = "2022-12-25", maxDate = "2023-02-25", keepZip = FALSE)
```

```r
# 1. Get all folders (you already did this)
folders <- prism_archive_ls()

# 2. Get full path to each .bil file
dl_dir <- prism_get_dl_dir()
bil_files <- file.path(dl_dir, folders, paste0(folders, ".bil"))

# 3. Load raster stack from actual .bil files
rasters <- stack(bil_files)

# 4. Define point of interestv

point <- SpatialPoints(cbind(-122.0308, 36.9741), 
                       proj4string = CRS("+proj=longlat +datum=WGS84"))

# 5. Extract values
values <- extract(rasters, point)

# 6. Extract dates from folder names
dates <- gsub("PRISM_ppt_stable_4kmD2_|_bil", "", basename(folders))
dates <- as.Date(dates, format = "%Y%m%d")

# 7. Build tidy data frame
data <- data.frame(Date = dates, Precipitation_mm = as.numeric(values))
data <- arrange(data, Date)
```
