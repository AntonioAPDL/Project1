```python
import subprocess
from datetime import datetime
from pytz import timezone

# Define conference deadlines
deadlines = {
    "ENAR": timezone('US/Pacific').localize(datetime(2023, 10, 17)),
    "JSM": timezone('US/Pacific').localize(datetime(2022, 12, 7)),
    "ICML": timezone('US/Pacific').localize(datetime(2023, 1, 26)),
    "GeoENV": None,
    "ISBA": timezone('US/Pacific').localize(datetime(2023, 11, 17)),
    "WNAR": timezone('US/Pacific').localize(datetime(2023, 11, 1)),
    # Add more conferences and deadlines here
    "NewConference": timezone('US/Pacific').localize(datetime(2023, 12, 31)),  # Example
}

# Function to calculate countdown
def calculate_countdown(deadline):
    current_time = datetime.now(timezone('US/Pacific'))  # Adjust timezone as needed
    delta = deadline - current_time
    return delta.days

# Function to generate LaTeX code with countdowns
def generate_latex():
    latex_code = r"""
    \documentclass{article}
    \usepackage{geometry}
    \usepackage{hyperref}
    \usepackage{multirow}
    \usepackage{tabularx}
    \usepackage{color}
    \usepackage{datetime}
    
    \geometry{a4paper, margin=1in}
    
    \newcolumntype{L}{>{\raggedright\arraybackslash}X}
    
    \definecolor{red}{RGB}{255,0,0}
    \definecolor{green}{RGB}{0,128,0}
    
    \begin{document}
    
    \begin{table}[h!]
    \centering
    \begin{tabularx}{\textwidth}{|l|L|L|l|}
    \hline
    \textbf{Conference} & \textbf{2023 Details} & \textbf{2024 Details} & \textbf{Countdown} \\
    \hline
    """

    for conference, deadline in deadlines.items():
        if deadline is None:
            countdown = "N/A"
        else:
            countdown = calculate_countdown(deadline)
            countdown = f"{countdown} days" if countdown >= 0 else "Deadline passed"
        
        latex_code += f"{conference} & & & {countdown} \\\\ \\hline\n"

    latex_code += r"""
    \end{tabularx}
    \caption{Important Statistics Conferences for 2023 and 2024}
    \end{table}
    
    \textbf{Legend:} \\
    \textcolor{red}{\textbf{Red}}: Deadlines before Dec 15, 2023 \\
    \textcolor{green}{\textbf{Green}}: Deadlines after Dec 15, 2023
    
    \end{document}
    """

    return latex_code

# Function to compile LaTeX code into PDF
def compile_latex(latex_code):
    with open('conference_deadlines.tex', 'w') as f:
        f.write(latex_code)

    subprocess.run(['pdflatex', 'conference_deadlines.tex'])
    subprocess.run(['pdflatex', 'conference_deadlines.tex'])  # Run twice for proper cross-referencing

if __name__ == "__main__":
    latex_code = generate_latex()
    compile_latex(latex_code)

```
