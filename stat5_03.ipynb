```python
import pandas as pd

# Load your data
sheet_url = 'https://docs.google.com/spreadsheets/d/1cv7bvwn7Kg7TxvOaleEcaM7e3SgEfM39G8K5xpk0yT8/export?gid=581234657&format=csv'
df = pd.read_csv(sheet_url)

# Step 1: Filter for D section only
d_section = 'D: Tu 08:30AM-09:35AM (Antonio)'
df_d = df[df['Section you attended:'] == d_section]

# Step 2: Group by email and aggregate name and count
result = (
    df_d
    .groupby('Email Address')
    .agg(
        Count=('Email Address', 'size')  
        )
    .reset_index()
)

# Show results
print(result)
```
