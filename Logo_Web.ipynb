```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Set up the figure
fig, ax = plt.subplots(figsize=(6, 6), dpi=300)

# Generate data for three overlapping density curves
x = np.linspace(-4, 4, 1000)
densities = [
    (norm.pdf(x, loc=-1, scale=1), '#1f77b4'),  # Blue
    (norm.pdf(x, loc=0, scale=0.8), '#ff7f0e'), # Orange
    (norm.pdf(x, loc=1.5, scale=0.6), '#2ca02c'), # Green
]

# Plot density curves
for density, color in densities:
    ax.fill_between(x, density, color=color, alpha=0.6)

# Add a text element for initials or a Gaussian symbol
ax.text(0, 0.35, r"$\mathcal{N}$", fontsize=70, fontweight="bold", color="black",
        ha='center', va='center')

# Remove axes for a cleaner look
ax.axis('off')

# Adjust layout for saving the image
plt.tight_layout()

# Save the logo
plt.savefig("stats_logo.png", transparent=True)
plt.show()
```
